```python
!pip install langchain-google-genai -q
!pip install -q google-generativeai

from langchain_google_genai import ChatGoogleGenerativeAI

# 1. Gemini 모델 설정 (원하는 모델명을 기입합니다)
# 기본적으로 환경 변수에 설정된 GOOGLE_API_KEY를 자동으로 인식합니다.
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")
```

```python
# @title 방법 1. from_template() 메소드를 사용하여 PromptTemplate 객체 생성
from langchain_core.prompts import PromptTemplate

# template 정의. {country}는 변수로, 이후에 값이 들어갈 자리를 의미
template = "{country}의 수도는 어디인가요?"

# from_template 메소드를 이용하여 PromptTemplate 객체 생성
prompt = PromptTemplate.from_template(template)
prompt
```

```python
# prompt 생성. format 메소드를 이용하여 변수에 값을 넣어줌
prompt = prompt.format(country="대한민국")
prompt
```

```python
# template 정의
template = "{country}의 수도는 어디인가요?"

# from_template 메소드를 이용하여 PromptTemplate 객체 생성
prompt = PromptTemplate.from_template(template)

# chain 생성
chain = prompt | llm
```

```python
# country 변수에 입력된 값이 자동으로 치환되어 수행됨
chain.invoke("대한민국").content
```

```python
# @title 방법 2. PromptTemplate 객체 생성과 동시에 prompt 생성
# template 정의
template = "{country}의 수도는 어디인가요?"

# PromptTemplate 객체를 활용하여 prompt_template 생성
prompt = PromptTemplate(
    template=template,
    input_variables=["country"],
)

prompt
```

```python
# prompt 생성
prompt.format(country="대한민국")
```

```python
# template 정의
template = "{country1}과 {country2}의 수도는 각각 어디인가요?"

# PromptTemplate 객체를 활용하여 prompt_template 생성
prompt = PromptTemplate(
    template=template,
    input_variables=["country1"],
    partial_variables={
        "country2": "미국"  # dictionary 형태로 partial_variables를 전달
    },
)

prompt
```

```python
prompt.format(country1="대한민국")
```

```python
prompt_partial = prompt.partial(country2="캐나다")
prompt_partial
```

```python
prompt_partial.format(country1="대한민국")
```

```python
chain = prompt_partial | llm
```

```python
chain.invoke("대한민국").content
```

```python
chain.invoke({"country1": "대한민국", "country2": "호주"}).content
```

```python
from datetime import datetime

# 오늘 날짜를 출력
datetime.now().strftime("%B %d")
```

```python
# 날짜를 반환하는 함수 정의
def get_today():
    return datetime.now().strftime("%B %d")
```

```python
prompt = PromptTemplate(
    template="오늘의 날짜는 {today} 입니다. 오늘이 생일인 유명인 {n}명을 나열해 주세요. 생년월일을 표기해주세요.",
    input_variables=["n"],
    partial_variables={
        "today": get_today  # dictionary 형태로 partial_variables를 전달
    },
)
```

```python
# prompt 생성
prompt.format(n=3)
```

```python
# chain 을 생성합니다.
chain = prompt | llm
```

```python
# chain 을 실행 후 결과를 확인합니다.
print(chain.invoke(3).content)
```

```python
# chain 을 실행 후 결과를 확인합니다.
print(chain.invoke({"today": "Jan 02", "n": 3}).content)
```

#### YAML 파일로 "프롬프트 관리의 분리와 모듈화"

- 파이썬 코드에 프롬프트를 하드코딩하지 않고 외부에 저장하면 다음과 같은 장점

- YAML 파일의 3가지 핵심 용도 및 장점

> **1. 프롬프트와 비즈니스 로직(코드)의 완전한 분리**
>> 프롬프트 문구를 수정할 때 파이썬 코드를 건드리지 않고 .yaml 파일만 수정

>> 개발자가 아닌 프롬프트 엔지니어나 기획자도 코드를 훼손할 위험 없이 프롬프트를 직관적으로 편집

> **2.프롬프트의 버전 관리 (Version Control)**

>> Git과 같은 버전 관리 시스템을 통해 프롬프트의 변경 이력을 개별 파일 단위로 쉽게 추적하고 관리

> **3. 재사용성 및 가독성 향상**

>> 길고 복잡한 프롬프트(System Prompt, 예시, 제약 조건 등)를 코드 내 삼중 따옴표("""...""")로 길게 작성할 필요 없이, 깔끔하게 외부 파일로 불러와 사용

```python
import os
# prompts 디렉터리 생성
os.makedirs("prompts", exist_ok=True)

# fruit_color.yaml 파일 작성
yaml_content = """_type: prompt
input_variables:
  - fruit
template: \"{fruit}의 색상은 무엇인가요?\"
"""

with open("prompts/fruit_color.yaml", "w", encoding="utf-8") as f:
    f.write(yaml_content)

print("prompts/fruit_color.yaml 파일이 성공적으로 생성되었습니다.")
```

```python
# @title 파일로부터 template 읽어오기
from langchain_core.prompts import load_prompt

prompt = load_prompt("prompts/fruit_color.yaml")
prompt
```

```python
prompt.format(fruit="사과")
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt = ChatPromptTemplate.from_template("{country}의 수도는 어디인가요?")
chat_prompt
```

```python
# @title ChatPromptTemplate
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages(
    [
        # role, message
        ("system", "당신은 친절한 AI 어시스턴트입니다. 당신의 이름은 {name} 입니다."),
        ("human", "반가워요!"),
        ("ai", "안녕하세요! 무엇을 도와드릴까요?"),
        ("human", "{user_input}"),
    ]
)

# 챗 message 를 생성합니다.
messages = chat_template.format_messages(
    name="조상구", user_input="당신의 이름은 무엇입니까?"
)

messages
```

```python
llm.invoke(messages).content
```

```python
chain = chat_template | llm
```

```python
chain.invoke({"name": "상구", "user_input": "당신의 이름은 무엇입니까?"}).content
```

```python
# @title MessagePlaceholder

from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

chat_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "당신은 요약 전문 AI 어시스턴트입니다. 당신의 임무는 주요 키워드로 대화를 요약하는 것입니다.",
        ),
        MessagesPlaceholder(variable_name="conversation"),
        ("human", "지금까지의 대화를 {word_count} 단어로 요약합니다."),
    ]
)
chat_prompt
```

```python
formatted_chat_prompt = chat_prompt.format(
    word_count=5,
    conversation=[
        ("human", "안녕하세요! 저는 오늘 새로 입사한 테디 입니다. 만나서 반갑습니다."),
        ("ai", "반가워요! 앞으로 잘 부탁 드립니다."),
    ],
)

print(formatted_chat_prompt)
```

```python
# chain 생성
chain = chat_prompt | llm | StrOutputParser()
```

```python
# chain 실행 및 결과확인
chain.invoke(
    {
        "word_count": 5,
        "conversation": [
            (
                "human",
                "안녕하세요! 저는 오늘 새로 입사한 조상구 입니다. 만나서 반갑습니다.",
            ),
            ("ai", "반가워요! 앞으로 잘 부탁 드립니다."),
        ],
    }
)
```
